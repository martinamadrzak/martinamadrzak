You must use the three functions outlined below in your solution. More functions can be added if you would like, but it is not necessary:

```python
#Complete the code in this cell

def print_border(width):

  #Complete the print statement
  print("*" * width)

def print_name_area(user_name, width):

  spaces = width - len(user_name) - 4

  #Finish the statements to print the correct number of spaces
  left_spaces = spaces // 2
  right_spaces = spaces // 2

  print("* " + " " * left_spaces + user_name + " " * right_spaces + " *")

def make_banner(user_name):

  width = len(user_name) + 8
#________________________________________________________________________
  #Fill in the arguments (the same argument may be required for different functions)
  print_border(width)
  print_name_area(user_name, width)
  print_border(width)

#Complete the line to prompt the user for their name
user_name = input("what is your name?")

make_banner(user_name)
```
